```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuración visual
sns.set_theme(style="whitegrid")

# Busca la carpeta datos tanto si corres desde cuadernos/ como desde la raíz
ruta = Path("../datos") if Path("../datos").exists() else Path("datos")

# Carga y consolidación de datos
train = pd.read_csv(ruta / "train.csv")
test = pd.read_csv(ruta / "test.csv")
df = pd.concat([train, test], ignore_index=True)
df.head()
```

**Punto 3**

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, col in zip(axes, ["Age", "Fare", "SibSp", "Parch"]):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

```python
for col in ["Age", "Fare", "SibSp", "Parch"]:
    q1, q3 = df[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    inf, sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n = ((df[col] < inf) | (df[col] > sup)).sum()
    print(f"{col}: límites [{inf:.2f}, {sup:.2f}] -> {n} outliers")
```

```python
df.sort_values("Fare", ascending=False)[["Name", "Pclass", "Fare", "Ticket"]].head(10)
plt.figure(figsize=(6, 4))
sns.boxplot(x="Pclass", y="Fare", data=df)
plt.show()
```

**Punto 5**

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["Title"] = df["Name"].str.extract(r" ([A-Za-z]+)\.", expand=False)
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Niño", "Adolescente", "Adulto joven", "Adulto", "Mayor"],
)
```
